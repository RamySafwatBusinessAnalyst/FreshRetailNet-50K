```python
import pandas as pd
import pyarrow.parquet as pq
from pathlib import Path

folder = Path(r"C:\Users\safwatr\IntelGraphicsProfiles\Fresh Retail")

parquet_file = folder / "train.parquet"
csv_file = folder / "train.csv"

pf = pq.ParquetFile(parquet_file)

first_batch = True
batch_size = 100_000

for i, batch in enumerate(pf.iter_batches(batch_size=batch_size)):
    df = batch.to_pandas()
    
    df.to_csv(
        csv_file,
        mode="w" if first_batch else "a",
        header=first_batch,
        index=False,
        encoding="utf-8-sig"
    )
    
    first_batch = False
    print(f"Batch {i+1} saved, rows: {len(df)}")

print("train.csv created successfully")
```
