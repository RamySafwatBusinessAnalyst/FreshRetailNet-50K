```python
import pandas as pd
from pathlib import Path

project_folder = Path(r"C:\Users\safwatr\IntelGraphicsProfiles\Fresh Retail")

train = pd.read_csv(project_folder / "train.csv")
eval_df = pd.read_csv(project_folder / "eval.csv")

print("Train shape:", train.shape)
print("Eval shape:", eval_df.shape)
```

```python
train["dt"] = pd.to_datetime(train["dt"])
eval_df["dt"] = pd.to_datetime(eval_df["dt"])

print(train["dt"].min(), train["dt"].max())
print(eval_df["dt"].min(), eval_df["dt"].max())
```

```python
from pathlib import Path

project_folder = Path(r"C:\Users\safwatr\IntelGraphicsProfiles\Fresh Retail")

raw_folder = project_folder
processed_folder = project_folder / "processed"
processed_folder.mkdir(exist_ok=True)

print("Project folder:", project_folder)
print("Processed folder:", processed_folder)
```

```python
daily_cols = [
    "city_id",
    "store_id",
    "management_group_id",
    "first_category_id",
    "second_category_id",
    "third_category_id",
    "product_id",
    "dt",
    "sale_amount",
    "stock_hour6_22_cnt",
    "discount",
    "holiday_flag",
    "activity_flag",
    "precpt",
    "avg_temperature",
    "avg_humidity",
    "avg_wind_level"
]

daily_train = train[daily_cols].copy()
daily_eval = eval_df[daily_cols].copy()

print("Daily train shape:", daily_train.shape)
print("Daily eval shape:", daily_eval.shape)

display(daily_train.head())
```

```python
daily_train.to_csv(
    processed_folder / "freshretail_daily_train.csv",
    index=False,
    encoding="utf-8-sig"
)

daily_eval.to_csv(
    processed_folder / "freshretail_daily_eval.csv",
    index=False,
    encoding="utf-8-sig"
)

print("Daily analytical CSV files saved successfully.")
print(processed_folder / "freshretail_daily_train.csv")
print(processed_folder / "freshretail_daily_eval.csv")
```

```python
metadata_summary = {
    "daily_train_rows": daily_train.shape[0],
    "daily_train_columns": daily_train.shape[1],
    "daily_eval_rows": daily_eval.shape[0],
    "daily_eval_columns": daily_eval.shape[1],
    "train_start_date": daily_train["dt"].min(),
    "train_end_date": daily_train["dt"].max(),
    "eval_start_date": daily_eval["dt"].min(),
    "eval_end_date": daily_eval["dt"].max(),
    "unique_cities": daily_train["city_id"].nunique(),
    "unique_stores": daily_train["store_id"].nunique(),
    "unique_products": daily_train["product_id"].nunique(),
    "management_groups": daily_train["management_group_id"].nunique(),
    "first_categories": daily_train["first_category_id"].nunique(),
    "second_categories": daily_train["second_category_id"].nunique(),
    "third_categories": daily_train["third_category_id"].nunique()
}

metadata_summary
```

```python
import pandas as pd

metadata_df = pd.DataFrame(
    list(metadata_summary.items()),
    columns=["metric", "value"]
)

metadata_df.to_csv(
    processed_folder / "freshretail_metadata_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

display(metadata_df)
```

```python
# Basic validation for prepared daily files

print("=== Shape Check ===")
print("Daily train:", daily_train.shape)
print("Daily eval:", daily_eval.shape)

print("\n=== Missing Values - Train ===")
missing_train = pd.DataFrame({
    "missing_count": daily_train.isna().sum(),
    "missing_percent": daily_train.isna().mean() * 100
}).sort_values("missing_count", ascending=False)

display(missing_train)

print("\n=== Duplicate Grain Check ===")
train_duplicates = daily_train.duplicated(
    subset=["store_id", "product_id", "dt"]
).sum()

eval_duplicates = daily_eval.duplicated(
    subset=["store_id", "product_id", "dt"]
).sum()

print("Train duplicate store-product-date rows:", train_duplicates)
print("Eval duplicate store-product-date rows:", eval_duplicates)

print("\n=== Date Range Check ===")
print("Train:", daily_train["dt"].min(), "to", daily_train["dt"].max())
print("Eval:", daily_eval["dt"].min(), "to", daily_eval["dt"].max())

print("\n=== Value Range Checks ===")
print("sale_amount min/max:", daily_train["sale_amount"].min(), daily_train["sale_amount"].max())
print("stock_hour6_22_cnt min/max:", daily_train["stock_hour6_22_cnt"].min(), daily_train["stock_hour6_22_cnt"].max())
print("discount min/max:", daily_train["discount"].min(), daily_train["discount"].max())

print("\n=== Flag Values ===")
print("holiday_flag values:", sorted(daily_train["holiday_flag"].unique()))
print("activity_flag values:", sorted(daily_train["activity_flag"].unique()))
```

```python
validation_summary = {
    "daily_train_rows": daily_train.shape[0],
    "daily_train_columns": daily_train.shape[1],
    "daily_eval_rows": daily_eval.shape[0],
    "daily_eval_columns": daily_eval.shape[1],
    "train_missing_values_total": daily_train.isna().sum().sum(),
    "eval_missing_values_total": daily_eval.isna().sum().sum(),
    "train_duplicate_store_product_date": daily_train.duplicated(
        subset=["store_id", "product_id", "dt"]
    ).sum(),
    "eval_duplicate_store_product_date": daily_eval.duplicated(
        subset=["store_id", "product_id", "dt"]
    ).sum(),
    "train_start_date": daily_train["dt"].min(),
    "train_end_date": daily_train["dt"].max(),
    "eval_start_date": daily_eval["dt"].min(),
    "eval_end_date": daily_eval["dt"].max(),
    "sale_amount_min": daily_train["sale_amount"].min(),
    "sale_amount_max": daily_train["sale_amount"].max(),
    "stock_hour6_22_cnt_min": daily_train["stock_hour6_22_cnt"].min(),
    "stock_hour6_22_cnt_max": daily_train["stock_hour6_22_cnt"].max(),
    "discount_min": daily_train["discount"].min(),
    "discount_max": daily_train["discount"].max(),
    "holiday_flag_values": str(sorted(daily_train["holiday_flag"].unique())),
    "activity_flag_values": str(sorted(daily_train["activity_flag"].unique()))
}

validation_df = pd.DataFrame(
    list(validation_summary.items()),
    columns=["check", "result"]
)

validation_df.to_csv(
    processed_folder / "freshretail_preparation_validation_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

display(validation_df)
```
