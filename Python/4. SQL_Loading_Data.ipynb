```python
from pathlib import Path
import pandas as pd

processed_folder = Path(r"C:\Users\safwatr\IntelGraphicsProfiles\Fresh Retail\processed")

eval_path = processed_folder / "freshretail_daily_eval.csv"

for i, chunk in enumerate(pd.read_csv(eval_path, chunksize=50000)):
    chunk["dt"] = pd.to_datetime(chunk["dt"]).dt.date
    
    chunk.to_sql(
        name="freshretail_daily_eval",
        con=engine,
        if_exists="append",
        index=False,
        chunksize=5000
    )
    
    print(f"Chunk {i+1} inserted, rows: {len(chunk)}")

print("Eval import completed")
```

```python
from pathlib import Path
import pandas as pd
import time

processed_folder = Path(r"C:\Users\safwatr\IntelGraphicsProfiles\Fresh Retail\processed")

train_path = processed_folder / "freshretail_daily_train.csv"

start_time = time.time()

for i, chunk in enumerate(pd.read_csv(train_path, chunksize=50000)):
    chunk["dt"] = pd.to_datetime(chunk["dt"]).dt.date
    
    chunk.to_sql(
        name="freshretail_daily_train",
        con=engine,
        if_exists="append",
        index=False,
        chunksize=5000
    )
    
    print(f"Chunk {i+1} inserted, rows: {len(chunk)}")

end_time = time.time()

print("Train import completed")
print("Total time minutes:", round((end_time - start_time) / 60, 2))
```
