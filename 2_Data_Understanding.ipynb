```python
import pandas as pd
from pathlib import Path

folder = Path(r"C:\Users\safwatr\IntelGraphicsProfiles\Fresh Retail")

train = pd.read_csv(folder / "train.csv")
eval_df = pd.read_csv(folder / "eval.csv")

print("Train shape:", train.shape)
print("Eval shape:", eval_df.shape)

print("\nTrain columns:")
print(train.columns.tolist())

print("\nTrain dtypes:")
print(train.dtypes)

display(train.head())
```

```python
# Convert dt to datetime
train["dt"] = pd.to_datetime(train["dt"])
eval_df["dt"] = pd.to_datetime(eval_df["dt"])

print("Date range:")
print("Train min date:", train["dt"].min())
print("Train max date:", train["dt"].max())
print("Eval min date:", eval_df["dt"].min())
print("Eval max date:", eval_df["dt"].max())

print("\nUnique counts in train:")
print("Cities:", train["city_id"].nunique())
print("Stores:", train["store_id"].nunique())
print("Products:", train["product_id"].nunique())
print("Management groups:", train["management_group_id"].nunique())
print("First categories:", train["first_category_id"].nunique())
print("Second categories:", train["second_category_id"].nunique())
print("Third categories:", train["third_category_id"].nunique())

print("\nGrain check:")
grain_duplicates = train.duplicated(subset=["store_id", "product_id", "dt"]).sum()
print("Duplicate rows at store-product-date grain:", grain_duplicates)

print("\nMissing values:")
missing_summary = pd.DataFrame({
    "missing_count": train.isna().sum(),
    "missing_percent": train.isna().mean() * 100
}).sort_values("missing_count", ascending=False)

display(missing_summary)
```
